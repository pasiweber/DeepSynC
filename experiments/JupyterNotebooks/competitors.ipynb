```python
import torch
import numpy as np
from clustpy.deep import DEC
```

```python
def load_optdigits():
    from clustpy.data.preprocessing import ZNormalizer
    from clustpy.data import load_optdigits as optdigits

    data_path = "/export/share/peters57dm/Verbund/data/benchmark"

    data, labels = optdigits(return_X_y=True, downloads_path=data_path)
    norm = ZNormalizer(False)
    normalized_data = norm.fit_transform(data)

    return normalized_data, labels
def detect_device():
    """Automatically detects if you have a cuda enabled GPU"""
    if torch.cuda.is_available():
        device = torch.device('cuda')
    else:
        device = torch.device('cpu')
    return device
```

```python
def evaluate_clustering(labels, clustering):
    from sklearn.metrics import normalized_mutual_info_score
    from sklearn.metrics import adjusted_rand_score
    from clustpy.metrics import unsupervised_clustering_accuracy
    """Calculates ACC, ARI and NMI for a given ground truth and clustering"""
    acc = unsupervised_clustering_accuracy(labels, clustering) * 100
    ari = adjusted_rand_score(labels, clustering) * 100
    nmi = normalized_mutual_info_score(labels, clustering) * 100
    print(f"ACC: {acc:.2f}, ARI: {ari:.2f}, NMI: {nmi:.2f}")
```

```python
batch_size = 256
device = detect_device()
clustering_n_epochs = 150
clustering_lr = 10e-4
random_state = 5
```

```python
import numpy as np
print(np.mean([0.6413017702671026, 0.5862611906267781, 0.7089718326449914]))
np.std([0.6413017702671026, 0.5862611906267781, 0.7089718326449914])
```

```python
data, gt_labels = load_optdigits()
true_n_clusters = len(np.unique(gt_labels))
```

```python
com = DEC(
        n_clusters=true_n_clusters,
        batch_size=batch_size,
        clustering_optimizer_params={'lr':clustering_lr},
        # neural_network=neural_network,
        clustering_epochs=clustering_n_epochs,
        device=device,
        random_state=random_state
    ).fit(data)
```

```python
evaluate_clustering(gt_labels, com.labels_)
```
